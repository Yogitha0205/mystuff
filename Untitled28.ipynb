```python
import json
import networkx as nx
import matplotlib.pyplot as plt

# Load the JSON data
with open(r"G:\CypherCop\JSON\test.json", "r") as f:
    data = json.load(f)

# Initialize a directed graph
G = nx.DiGraph()

# Track total received BTC for node sizing
address_totals = {}

# Loop through transactions
for tx in data.get("txs", []):
    txid = tx["hash"]

    input_addresses = []
    for vin in tx.get("inputs", []):
        if "addresses" in vin and vin["addresses"]:
            input_addresses.extend(vin["addresses"])

    outputs = []
    for vout in tx.get("outputs", []):
        if "addresses" in vout and vout["addresses"]:
            for addr in vout["addresses"]:
                btc_value = vout.get("value", 0) / 100000000
                outputs.append((addr, btc_value))
                address_totals[addr] = address_totals.get(addr, 0) + btc_value

    for inp in input_addresses:
        for out_addr, btc_value in outputs:
            label = f"{btc_value:.4f} BTC\n{txid[:8]}..."
            G.add_edge(inp, out_addr, label=label, weight=btc_value)

# Node sizes proportional to BTC received
node_sizes = [2000 + address_totals.get(node, 0) * 200 for node in G.nodes()]

# Positions: use shell layout for more clarity
pos = nx.shell_layout(G)

plt.figure(figsize=(16, 12))

# Draw nodes
nx.draw_networkx_nodes(G, pos, node_size=node_sizes, node_color="skyblue", alpha=0.9)

# Draw edges (thicker for higher amounts)
edge_weights = [1 + G[u][v]["weight"]*2 for u,v in G.edges()]
nx.draw_networkx_edges(G, pos, width=edge_weights, edge_color="gray", arrows=True, connectionstyle="arc3,rad=0.1")

# Draw labels
nx.draw_networkx_labels(G, pos, font_size=8, font_weight="bold")

# Edge labels
edge_labels = nx.get_edge_attributes(G, "label")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=6, rotate=False)

plt.title("Bitcoin Address Flow Graph (Weighted + Cleaner Layout)", fontsize=14)
plt.axis("off")
plt.show()
```
