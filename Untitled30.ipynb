```python
import requests
import networkx as nx
import matplotlib.pyplot as plt

# Example wallet with visible activity
WALLET = "1dice8EMZmqKvrGE4Qc9bUFf9PX3xaYDp"

def fetch_txs(wallet, limit=20):
    url = f"https://api.blockcypher.com/v1/btc/main/addrs/{wallet}/full?limit={limit}"
    r = requests.get(url)
    txs = r.json().get("txs", [])
    print(f"Fetched {len(txs)} transactions from wallet {wallet}")
    return txs

# Build graph
G = nx.DiGraph()
txs = fetch_txs(WALLET, limit=20)

for tx in txs:
    txid = tx.get("hash", "unknown")
    inputs = [a for vin in tx.get("inputs", []) for a in (vin.get("addresses") or [])]
    outputs = [(a, vout.get("value", 0)) for vout in tx.get("outputs", []) for a in (vout.get("addresses") or [])]
    for src in inputs:
        for dst, val in outputs:
            btc = val / 1e8
            G.add_edge(src, dst, label=f"{btc:.4f} BTC\n{txid[:6]}")

print(f"Graph nodes: {len(G.nodes)}, edges: {len(G.edges)}")

# Detect suspicious nodes
suspicious = set()
for tx in txs:
    ins = [a for vin in tx.get("inputs", []) for a in vin.get("addresses") or []]
    outs = [a for vout in tx.get("outputs", []) for a in vout.get("addresses") or []]
    if len(ins) == 1 and len(outs) >= 5:
        suspicious.update(outs)  # fan-out
    if len(ins) >= 5 and len(outs) == 1:
        suspicious.update(ins)   # fan-in
    if len(ins) == 1 and len(outs) == 2:
        suspicious.update(outs)  # peeling chain

# Graphviz layout for clarity (requires Graphviz installed)
try:
    pos = nx.nx_agraph.graphviz_layout(G, prog="dot")
except:
    pos = nx.spring_layout(G)

# Node coloring
colors = []
for node in G.nodes:
    if node == WALLET:
        colors.append("red")
    elif node in suspicious:
        colors.append("orange")
    else:
        colors.append("lightblue")

# Plot
plt.figure(figsize=(20, 15))
nx.draw(G, pos, with_labels=False, node_color=colors, node_size=800, arrowsize=15)
nx.draw_networkx_labels(G, pos, font_size=8)
edge_labels = nx.get_edge_attributes(G, "label")
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=6)
plt.title("Transaction Flow\nRed = Tracked Wallet | Orange = Suspicious Patterns", fontsize=16)
plt.savefig("suspicious_flow.png", dpi=300, bbox_inches="tight")
plt.show()
print("Graph saved as suspicious_flow.png")
```
