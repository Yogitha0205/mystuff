```python
# Bitcoin cluster notebook - more robust fetch + debugging
# Requirements: requests, networkx, matplotlib
import requests
import networkx as nx
import matplotlib.pyplot as plt
from pprint import pprint
import time

# -----------------------------
# CONFIG
# -----------------------------
# Replace with your seed; if you had errors, try the example below:
# WORKING_EXAMPLE = "1BoatSLRHtKNngkdXEeobR76b53LETtpyT"
SEED_ADDRESS = "1BoatSLRHtKNngkdXEeobR76b53LETtpyT"
API_BASE = "https://api.blockcypher.com/v1/btc/main"
# Optional: your BlockCypher token (get one from blockcypher.com if you hit rate limits/403s)
BLOCKCYPHER_TOKEN = None  # e.g. "your_token_here"

# -----------------------------
# HELPERS - robust fetch + fallbacks
# -----------------------------
def _get(url, params=None, max_retries=3, backoff=1.0):
    params = params or {}
    for attempt in range(1, max_retries+1):
        try:
            r = requests.get(url, params=params, timeout=20)
            return r
        except requests.RequestException as e:
            if attempt == max_retries:
                raise
            time.sleep(backoff * attempt)

def fetch_transactions(address, limit=50, try_full_if_empty=True, token=None):
    """
    Fetch transaction references for an address using BlockCypher.
    Tries the lightweight endpoint first (/addrs/{address}) then falls back to /addrs/{address}/full.
    Returns a list of txref-like dicts with fields: tx_hash, address, tx_output_n, value
    """
    params = {"limit": limit}
    if token:
        params["token"] = token

    url_shallow = f"{API_BASE}/addrs/{address}"
    r = _get(url_shallow, params=params)
    # helpful debug if something goes wrong
    if r.status_code != 200:
        # show status + body for debugging
        raise RuntimeError(f"BlockCypher returned HTTP {r.status_code} for {url_shallow}\nResponse body:\n{r.text}")

    data = r.json()

    # gather txrefs if present (confirmed + unconfirmed)
    txrefs = []
    if "txrefs" in data and data["txrefs"]:
        txrefs.extend(data["txrefs"])
    if "unconfirmed_txrefs" in data and data["unconfirmed_txrefs"]:
        txrefs.extend(data["unconfirmed_txrefs"])

    # If we found txrefs, normalize and return
    if txrefs:
        # Normalize fields we use later (ensure keys exist)
        normalized = []
        for t in txrefs:
            normalized.append({
                "tx_hash": t.get("tx_hash") or t.get("txid") or t.get("tx_hash"),
                "address": t.get("address"),
                "tx_output_n": t.get("tx_output_n", t.get("tx_output_n")),
                "value": t.get("value", 0)
            })
        return normalized

    # If no txrefs and fallback allowed, try full endpoint which returns transactions with inputs/outputs
    if try_full_if_empty:
        url_full = f"{API_BASE}/addrs/{address}/full"
        r2 = _get(url_full, params=params)
        if r2.status_code != 200:
            raise RuntimeError(f"BlockCypher returned HTTP {r2.status_code} for {url_full}\nResponse body:\n{r2.text}")
        d2 = r2.json()
        txs = d2.get("txs") or d2.get("transactions") or []
        if not txs:
            # no txs found
            return []
        # convert txs to txref-like records
        normalized = []
        for tx in txs:
            txid = tx.get("hash") or tx.get("tx_hash") or tx.get("txid")
            # inputs
            for vin in tx.get("inputs", []) or []:
                # vin addresses may be a list under 'addresses'
                addrs = vin.get("addresses") or vin.get("address") or []
                if isinstance(addrs, str):
                    addrs = [addrs]
                for a in addrs:
                    normalized.append({
                        "tx_hash": txid,
                        "address": a,
                        "tx_output_n": -1,  # denote input
                        "value": vin.get("output_value", 0) or vin.get("value", 0) or 0
                    })
            # outputs
            for idx, vout in enumerate(tx.get("outputs", []) or []):
                addrs = vout.get("addresses") or vout.get("address") or []
                if isinstance(addrs, str):
                    addrs = [addrs]
                for a in addrs:
                    normalized.append({
                        "tx_hash": txid,
                        "address": a,
                        "tx_output_n": idx,
                        "value": vout.get("value", 0) or 0
                    })
        return normalized

    # nothing found
    return []

# -----------------------------
# GRAPH + CLUSTERING (same as before)
# -----------------------------
def build_graph(txrefs):
    G = nx.DiGraph()
    for tx in txrefs:
        txid = tx["tx_hash"]
        addr = tx["address"]
        val = tx.get("value", 0)

        # tx node
        G.add_node(txid, type="tx")
        G.add_node(addr, type="addr")

        if tx.get("tx_output_n", 0) >= 0:
            # tx -> addr
            G.add_edge(txid, addr, kind="output", value=val)
        else:
            # addr -> tx
            G.add_edge(addr, txid, kind="input", value=val)
    return G

def common_input_clusters(txrefs):
    cluster_graph = nx.Graph()
    tx_inputs = {}
    for tx in txrefs:
        txid = tx["tx_hash"]
        addr = tx["address"]
        if tx.get("tx_output_n", 0) < 0:
            tx_inputs.setdefault(txid, []).append(addr)
    for addrs in tx_inputs.values():
        for i in range(len(addrs)):
            for j in range(i + 1, len(addrs)):
                cluster_graph.add_edge(addrs[i], addrs[j])
    return list(nx.connected_components(cluster_graph))

# -----------------------------
# RUN
# -----------------------------
try:
    txrefs = fetch_transactions(SEED_ADDRESS, limit=100, try_full_if_empty=True, token=BLOCKCYPHER_TOKEN)
    print(f"Fetched {len(txrefs)} txrefs for {SEED_ADDRESS}")
    if not txrefs:
        print("No transactions found for this address. Try a different address or testnet address.")
    else:
        G = build_graph(txrefs)
        clusters = common_input_clusters(txrefs)
        print("Clusters detected (common-input):")
        for c in clusters:
            print(c)

        # small visualization (works for modest graph sizes)
        plt.figure(figsize=(10,7))
        pos = nx.spring_layout(G, seed=42)
        addr_nodes = [n for n,d in G.nodes(data=True) if d.get("type")=="addr"]
        tx_nodes = [n for n,d in G.nodes(data=True) if d.get("type")=="tx"]
        nx.draw_networkx_nodes(G, pos, nodelist=addr_nodes, node_size=200, alpha=0.8)
        nx.draw_networkx_nodes(G, pos, nodelist=tx_nodes, node_size=80, alpha=0.6)
        nx.draw_networkx_edges(G, pos, arrows=False, alpha=0.3)
        # draw a subset of labels (addresses only) to avoid clutter
        labels = {n:n for n in addr_nodes[:15]}
        nx.draw_networkx_labels(G, pos, labels, font_size=7)
        plt.title(f"Graph for {SEED_ADDRESS}")
        plt.axis("off")
        plt.show()

except Exception as e:
    print("ERROR:", str(e))
    # If it's helpful: show troubleshooting hints below
    print("\nTroubleshooting hints:")
    print("- If BlockCypher responded with HTTP 403 or 429: add a BLOCKCYPHER_TOKEN and retry.")
    print("- If you get HTTP 404: check the address string is valid and that you're using the correct chain (mainnet vs testnet).")
    print("- If response body shows a JSON error from BlockCypher, paste that JSON into your next message so I can interpret it.")
    print("- To test locally, try the included example address which normally has transactions.")
```
