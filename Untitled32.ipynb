```python
import requests
import time

# Blockchair API endpoint for Bitcoin transactions
BASE_URL = "https://api.blockchair.com/bitcoin/transactions"

# The current Bitcoin price (use a reliable source like CoinGecko, CoinMarketCap, or a live API to fetch it dynamically)
BTC_TO_USD = 20000  # Adjust with current BTC price
MIN_TRANSACTION_VALUE = 10000  # USD

# Function to fetch and filter transactions over $10,000 with retry on rate limit
def get_high_value_transactions_with_retry():
    params = {
        'limit': 100,
        'fields': 'transaction_hash,total_sum,input_addresses,output_addresses'
    }

    def fetch_data_with_retry(url, params, retries=3, delay=5):
        for i in range(retries):
            response = requests.get(url, params=params)
            if response.status_code == 200:
                return response.json()
            elif response.status_code == 430:
                print(f"Rate limit hit. Retrying in {delay} seconds...")
                time.sleep(delay)
            else:
                print(f"Error fetching data: {response.status_code}")
                break
        return None

    data = fetch_data_with_retry(BASE_URL, params)
    if data:
        transactions = data['data']
        print(f"Found {len(transactions)} transactions.")
        
        high_value_transactions = []
        for tx in transactions:
            total_sum_satoshis = tx['total_sum']
            total_sum_btc = total_sum_satoshis / 100000000
            total_value_usd = total_sum_btc * BTC_TO_USD
            
            if total_value_usd > MIN_TRANSACTION_VALUE:
                high_value_transactions.append({
                    'transaction_hash': tx['transaction_hash'],
                    'total_sum_btc': total_sum_btc,
                    'total_value_usd': total_value_usd,
                    'input_addresses': tx['input_addresses'],
                    'output_addresses': tx['output_addresses']
                })

        if high_value_transactions:
            print("High-value transactions (greater than $10,000):")
            for tx in high_value_transactions:
                print(f"Transaction Hash: {tx['transaction_hash']}")
                print(f"Total Amount (BTC): {tx['total_sum_btc']}")
                print(f"Total Value (USD): ${tx['total_value_usd']:.2f}")
                print(f"Input Addresses: {tx['input_addresses']}")
                print(f"Output Addresses: {tx['output_addresses']}")
                print("-" * 50)
        else:
            print("No transactions found greater than $10,000.")
    else:
        print("Failed to fetch transactions after retrying.")

# Call the function to get high-value transactions
get_high_value_transactions_with_retry()
```
