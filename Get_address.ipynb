```python
import requests

def get_recent_addresses(limit=5):
    base_url = "https://api.blockcypher.com/v1/btc/main"
    
    try:
        # 1. Get latest block
        latest_block = requests.get(base_url).json()
        block_hash = latest_block.get("hash")
        print(f"📦 Latest block hash: {block_hash}")
        
        # 2. Get block details
        block_data = requests.get(f"{base_url}/blocks/{block_hash}").json()
        txids = block_data.get("txids", [])
        
        if not txids:
            print("⚠️ No transactions found in latest block.")
            return []
        
        print(f"🔎 Found {len(txids)} transactions in latest block.")
        
        addresses = set()
        
        # 3. Look into first few transactions
        for txid in txids[:limit]:
            tx_data = requests.get(f"{base_url}/txs/{txid}").json()
            
            # Extract input addresses (if present)
            for inp in tx_data.get("inputs", []):
                if inp.get("addresses"):  # check before looping
                    for addr in inp["addresses"]:
                        addresses.add(addr)
            
            # Extract output addresses (if present)
            for out in tx_data.get("outputs", []):
                if out.get("addresses"):
                    for addr in out["addresses"]:
                        addresses.add(addr)
        
        return list(addresses)
    
    except Exception as e:
        print("⚠️ Error fetching data:", e)
        return []

# Example usage
real_addresses = get_recent_addresses(limit=5)
print("\n✅ Recent real Bitcoin addresses found:")
for addr in real_addresses:
    print("-", addr)
```
