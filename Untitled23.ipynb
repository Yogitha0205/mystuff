```python
# ---
# Title: Bitcoin Wallet Cluster Detection Prototype
# Author: ChatGPT
# ---

import requests
import networkx as nx
import matplotlib.pyplot as plt

# -----------------------------
# CONFIG
# -----------------------------
# Choose a Bitcoin wallet address as a seed.
# For demo: we'll use a random testnet or known Bitcoin address.
SEED_ADDRESS = "1DEP8i3QJCsomS4BSMY2RpU1upv62aGvhD"  # Example from BlockCypher docs
API_BASE = "https://api.blockcypher.com/v1/btc/main"

# -----------------------------
# STEP 1: Fetch transactions for the seed address
# -----------------------------
def fetch_transactions(address, limit=20):
    #"""Fetch recent transactions for an address from BlockCypher API."""
    url = f"{API_BASE}/addrs/{address}?limit={limit}&includeHex=false"
    resp = requests.get(url)
    if resp.status_code != 200:
        raise Exception(f"Error fetching data: {resp.text}")
    data = resp.json()
    return data.get("txrefs", []), data.get("address")

# -----------------------------
# STEP 2: Build transaction graph
# -----------------------------
def build_graph(txrefs):
    #"""Build directed graph of tx -> address edges."""
    G = nx.DiGraph()
    for tx in txrefs:
        txid = tx["tx_hash"]
        addr = tx["address"]
        val = tx["value"]

        # Add tx node and edge (address <-> tx)
        G.add_node(txid, type="tx")
        G.add_node(addr, type="addr")

        if tx["tx_output_n"] >= 0:
            # Output edge: tx -> addr
            G.add_edge(txid, addr, kind="output", value=val)
        else:
            # Input edge: addr -> tx
            G.add_edge(addr, txid, kind="input", value=val)
    return G

# -----------------------------
# STEP 3: Apply common-input heuristic
# -----------------------------
def common_input_clusters(txrefs):
    #"""Cluster addresses that appear as inputs to the same tx."""
    cluster_graph = nx.Graph()
    tx_inputs = {}

    for tx in txrefs:
        txid = tx["tx_hash"]
        addr = tx["address"]
        if tx["tx_output_n"] < 0:  # it's an input
            tx_inputs.setdefault(txid, []).append(addr)

    # Connect input addresses pairwise
    for addrs in tx_inputs.values():
        for i in range(len(addrs)):
            for j in range(i + 1, len(addrs)):
                cluster_graph.add_edge(addrs[i], addrs[j])

    return list(nx.connected_components(cluster_graph))

# -----------------------------
# STEP 4: Run the pipeline
# -----------------------------
txrefs, seed = fetch_transactions(SEED_ADDRESS, limit=50)
print(f"Fetched {len(txrefs)} txrefs for {seed}")

G = build_graph(txrefs)
clusters = common_input_clusters(txrefs)

print("Clusters detected:")
for c in clusters:
    print(c)

# -----------------------------
# STEP 5: Visualize small graph
# -----------------------------
plt.figure(figsize=(10,7))
pos = nx.spring_layout(G, seed=42)
addr_nodes = [n for n,d in G.nodes(data=True) if d["type"]=="addr"]
tx_nodes = [n for n,d in G.nodes(data=True) if d["type"]=="tx"]

nx.draw_networkx_nodes(G, pos, nodelist=addr_nodes, node_color="skyblue", node_size=400, label="Addresses")
nx.draw_networkx_nodes(G, pos, nodelist=tx_nodes, node_color="orange", node_size=300, label="Transactions")
nx.draw_networkx_edges(G, pos, arrows=True, alpha=0.4)
nx.draw_networkx_labels(G, pos, font_size=8)

plt.legend()
plt.title(f"Graph for seed address {SEED_ADDRESS}")
plt.axis("off")
plt.show()
```
