```python
import json
import networkx as nx
import matplotlib.pyplot as plt

# Load the JSON data (replace with your actual filename or API response)
with open(r"G:\CypherCop\JSON\test.json", "r") as f:
    data = json.load(f)

# Initialize a directed graph
G = nx.DiGraph()

# Loop through each transaction
for tx in data.get("txs", []):
    txid = tx["hash"]

    # Collect all input addresses
    input_addresses = []
    for vin in tx.get("inputs", []):
        if "addresses" in vin and vin["addresses"]:
            input_addresses.extend(vin["addresses"])

    # Collect all output addresses
    output_addresses = []
    for vout in tx.get("outputs", []):
        if "addresses" in vout and vout["addresses"]:
            output_addresses.extend(vout["addresses"])

    # Add edges: each input -> each output
    for inp in input_addresses:
        for out in output_addresses:
            G.add_edge(inp, out, txid=txid)

# Draw the graph
plt.figure(figsize=(12, 8))
pos = nx.spring_layout(G, k=0.5, iterations=50)

nx.draw(G, pos, with_labels=True, node_size=1500, node_color="lightblue", font_size=8, arrows=True)
nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, "txid"), font_size=6)

plt.title("Bitcoin Address Flow Graph")
plt.show()
```
