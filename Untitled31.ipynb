```python
import requests
import networkx as nx
import matplotlib.pyplot as plt

wallet_address = "1KbrSKx5yRtC9Z2D4XpXzjCZUs1uM6fL5t"
url = f"https://api.blockcypher.com/v1/btc/main/addrs/{wallet_address}/full?limit=200"
resp = requests.get(url)
data = resp.json()

print("Keys in API response:", list(data.keys()))
print("Number of transactions returned:", len(data.get("txs", [])))

if len(data.get("txs", [])) == 0:
    print("⚠️ No transactions fetched — try another wallet or API is rate-limiting.")
else:
    print("✅ Transactions available, building graph...")

    # build simple graph to confirm
    G = nx.DiGraph()
    for tx in data.get("txs", []):
        txid = tx["hash"]
        for inp in tx.get("inputs", []):
            addr = inp.get("addresses", ["unknown"])[0]
            G.add_edge(addr, txid)
        for out in tx.get("outputs", []):
            addr = out.get("addresses", ["unknown"])[0]
            G.add_edge(txid, addr)

    print("Graph nodes:", len(G.nodes))
    print("Graph edges:", len(G.edges))

    pos = nx.spring_layout(G, seed=42, k=0.5)
    plt.figure(figsize=(12, 8))
    nx.draw(G, pos, with_labels=False, node_color="lightblue", node_size=600)
    plt.title(f"Transaction Graph for {wallet_address}")
    plt.savefig("wallet_graph.png", dpi=300)
    plt.show()
    print("✅ Graph saved as wallet_graph.png")
```
